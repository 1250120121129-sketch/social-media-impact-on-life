```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df=pd.read_csv("C:/Users/acer/Downloads/Social_media_impact_on_life.csv")
```

```python
df
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Daily_Usage_Hours"], bins=15, edgecolor="black")

plt.title("Daily Social Media Usage")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

df["Overall_Impact"].value_counts().plot(kind="bar")

plt.title("Overall Impact of Social Media")
plt.xlabel("Overall Impact")
plt.ylabel("Count")

plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Daily_Usage_Hours"],
    df["Academic_Performance_GPA"]
)

plt.title("Daily Usage vs Academic Performance")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Academic Performance GPA")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Daily_Usage_Hours"],
    df["Sleep_Duration_Hours"]
)

plt.title("Daily Usage vs Sleep Duration")
plt.xlabel("Daily Usage Hours")
plt.ylabel("Sleep Duration Hours")

plt.show()
```
